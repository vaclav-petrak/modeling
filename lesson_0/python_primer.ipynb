### 1. Python Basics: Spacing and Indentation
Unlike many other languages (like C++ or Java) that use curly braces {} to group code, Python uses **indentation** (whitespace). 

- **Whitespace Matters:** You must indent code inside loops, functions, or if-statements. 
- **Standard:** The standard is 4 spaces (or 1 tab).
- **Comments:** Lines starting with `#` are ignored by the computer.

### 2. Python vs. MATLAB
If you are coming from MATLAB, here are the critical differences to remember:

1. **Indexing:** Python starts counting at **0**. MATLAB starts at **1**.
   - MATLAB: `vector(1)` is the first item.
   - Python: `vector[0]` is the first item.
   
2. **Brackets:** - Python uses square brackets `[]` for lists/indexing and parentheses `()` for function calls.
   - MATLAB uses parentheses `()` for almost everything.

3. **Powers:** - Python uses `**` for exponents (e.g., `2**3`).
   - MATLAB uses `^` (e.g., `2^3`).

### 5. Lists (Collections)
A 'list' is a container that holds multiple items in a specific order. You often use lists when you want to collect data one step at a time, such as recording a measurement every day.

- Lists use square brackets [ ].
- Indices start at 0 (the first item is at index 0).
- You can add items to the end of a list using .append().

```python
# Creating an empty list
data_points = []

# Adding values one by one
data_points.append(5.5)
data_points.append(6.2)
data_points.append(7.1)

print(data_points)
# Output: [5.5, 6.2, 7.1]

# Accessing specific items
first_value = data_points[0]  # Get 5.5
last_value = data_points[-1]  # Get 7.1 (Negative numbers count from the end)
```

#### Pre-allocation vs. Appending
In simulations, we often need a list to store results.

Method A: Pre-allocation (Faster, Fixed Size)
If you know you need to simulate 100 time steps, create a list of 100 zeros first. This is computationally more efficient.

Method B: Appending (Flexible)
If you don't know how many steps you need (e.g., "run until the tank is empty"), start with an empty list and add to it.

```python
# --- Method A: Pre-allocation ---
steps = 5
# Create a list [0, 0, 0, 0, 0]
history = [0] * steps 

for i in range(steps):
    history[i] = i * 2  # Fill specific slot
print(history) 
# Output: [0, 2, 4, 6, 8]


# --- Method B: Appending ---
values = [] # Start empty
for i in range(steps):
    values.append(i * 2) # Add to end
print(values)
# Output: [0, 2, 4, 6, 8]
```

### 6. Loops (Repetition)
Computers excel at doing the same thing over and over. In modeling, we use loops to step through time (e.g., "for every year from 0 to 10...").

There are two main types of loops:
1. For Loop: Use this when you know exactly how many times you want to repeat something.
2. While Loop: Use this when you want to repeat something until a specific condition changes.

```python
# --- Example A: The FOR Loop ---
# Calculate the square of numbers 0 through 4
# range(5) creates a sequence: 0, 1, 2, 3, 4

squares = []
for i in range(5):
    val = i * i
    squares.append(val)

print("Squares:", squares)


# --- Example B: The WHILE Loop ---
# Keep doubling a number until it exceeds 100

current_value = 1
while current_value < 100:
    print(current_value)
    current_value = current_value * 2

print("Final value:", current_value)
```

### 7. NumPy Arrays (Vectorized Math)
While standard lists are good for collecting data, they are bad for doing math. If you try to multiply a Python list by 2, it duplicates the list instead of multiplying the numbers inside.

For mathematical modeling, we use the library 'NumPy'. It creates "Arrays" that behave like mathematical vectors. You can multiply, add, or divide entire arrays in a single line of code without writing a loop.

```python
import numpy as np

# A standard Python list
my_list = [1, 2, 3]
# print(my_list * 2) -> Output would be [1, 2, 3, 1, 2, 3] (Not math!)

# A NumPy Array
my_array = np.array([1, 2, 3])

# Now math works on every element at once
doubled = my_array * 2
print(doubled)


```

 You can also do math between two arrays (e.g., calculating distance = speed * time)

```python
speed = np.array([10, 20, 30])
time  = np.array([ 2,  2,  2])

distance = speed * time
print(distance)
```

### 8. Slicing (Selecting Data)
When analyzing data, you often need to look at just a specific section (e.g., "The first 10 years" or "Every value greater than zero"). NumPy allows you to "slice" arrays easily.

```python
import numpy as np

data = np.array([10, 20, 30, 40, 50, 60])

# Syntax: [start : end]
# Note: Python stops *before* the 'end' index.
subset = data[1:4]
print(subset)
# Output: [20 30 40]

# Conditional Indexing (Masking)
# Select only values greater than 35
high_values = data[data > 35]
print(high_values)
# Output: [40 50 60]
```

### 9. Plotting
Visualizing your model is critical. We use the library 'matplotlib.pyplot' to draw graphs. The standard workflow is:
1. Create your x-axis data (e.g., Time).
2. Create your y-axis data (e.g., Value).
3. Pass them to plt.plot().

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Generate Data
x = np.linspace(0, 10, 100)  # Create 20 points between 0 and 10
y = np.sin(x)               # Calculate Sine of x

# 2. Setup Plot
plt.figure(figsize=(8, 4))  # Optional: Set size

# 3. Plot the line
# 'b-' means blue solid line, 'ro' would mean red dots
plt.plot(x, y, 'b-', label='Sine Wave')

# 4. Add Labels
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.title("Simple Oscillation Example")
plt.legend()
plt.grid(True)

# 5. Display
plt.show()
```

### 10. Text Formatting (f-strings)
When printing results, you often want to mix text and numbers or round decimal places. Python uses **f-strings** (formatted string literals) for this. Put an `f` before the quotes and use `{}` to insert variables.

```python
name = "Sensor A"
value = 23.456789

# Basic formatting
print(f"The reading from {name} is {value}")
# Output: The reading from Sensor A is 23.456789

# Rounding numbers
# :.2f means "Float with 2 decimal places"
print(f"Adjusted Value: {value:.2f}")
# Output: Adjusted Value: 23.46
```

### 11. Matrix Operations (NumPy)
For 2D data (like a grid of values), we use NumPy arrays. 
- **Shape:** Rows x Columns.
- **Slicing:** Access sub-sections easily.

```python
import numpy as np

# Create a 2x3 Matrix
# Row 0: [10, 20, 30]
# Row 1: [40, 50, 60]
data = np.array([[10, 20, 30], [40, 50, 60]])

# Accessing a single value [Row, Column]
print(data[0, 2])  # Row 0, Col 2 -> Output: 30

# Slicing a whole column
# [ : , 1 ] means "All rows, Column 1"
print(data[:, 1])  # Output: [20 50]

# Element-wise Math (No loop needed!)
# Multiply every number in the matrix by 0.1
print(data * 0.1)
# Output: [[1. 2. 3.] [4. 5. 6.]]
```

### 12. Functions
Functions allow you to save a specific task and reuse it. 
- **def:** Keyword to define a function.
- **Arguments:** Inputs go in the parentheses.
- **return:** Output sent back to the main program.

```python
# Define a function to calculate the area of a rectangle
def calculate_area(length, width):
    area = length * width
    return area

# Now we can reuse it for different objects without rewriting the math
room_a = calculate_area(10, 5)
room_b = calculate_area(7, 3)

print("Room A Area:", room_a)
print("Room B Area:", room_b)
```

A function can also return a list (e.g., a sequence generator)

```python
def generate_evens(limit):
    evens = []
    for i in range(limit):
        if i % 2 == 0:
            evens.append(i)
    return evens

numbers = generate_evens(10)
print("Evens:", numbers)
```
